```python
import numpy as np
import matplotlib.pyplot as plt

nu = 0.6
VM = 300 
VT = VM * nu
K_values = [1, 5, 10]  # Gain for part (A)
T_max = 5
dt = 0.01
g = 9.81
aT_const = 3*g   # use same name everywhere

# Initial positions and headings
xM0, yM0 = 0, 0
xT0, yT0 = 4000, 1000
thetaM0 = np.deg2rad(0)
thetaT0 = np.deg2rad(180)

# ---------- Part (A) ----------
results_A = []
for K in K_values:
    # Initialize
    xM, yM = xM0, yM0
    xT, yT = xT0, yT0
    thetaM = thetaM0
    thetaT = thetaT0
    t = 0

    RM_list, theta_diff_list, aM_list, t_list = [], [], [], []
    xM_list, yM_list, xT_list, yT_list = [], [], [], []

    while t < T_max:
        dx = xT - xM
        dy = yT - yM
        RM = np.hypot(dx, dy)
        theta_diff = (thetaT - thetaM)
        theta_diff = (theta_diff + np.pi) % (2 * np.pi) - np.pi

        # Missile lateral acceleration
        aM = K * RM * theta_diff

        # Update headings
        thetaM += (aM / VM) * dt

        # Update positions
        xM += VM * np.cos(thetaM) * dt
        yM += VM * np.sin(thetaM) * dt
        xT += VT * np.cos(thetaT) * dt
        yT += VT * np.sin(thetaT) * dt

        # Store for plotting
        RM_list.append(RM)
        theta_diff_list.append(RM * theta_diff)
        aM_list.append(aM)
        t_list.append(t)
        xM_list.append(xM)
        yM_list.append(yM)
        xT_list.append(xT)
        yT_list.append(yT)

        if RM < 1:
            break

        t += dt

    miss_distance = RM_list[-1]
    print(f"(A) K={K}: Miss distance = {miss_distance:.2f} m")

    results_A.append({
        "t": np.array(t_list),
        "aM": np.array(aM_list),
        "RM_theta_diff": np.array(theta_diff_list),
        "RM": np.array(RM_list),
        "xM": np.array(xM_list),
        "yM": np.array(yM_list),
        "xT": np.array(xT_list),
        "yT": np.array(yT_list),
        "miss_distance": miss_distance
    })

# --- Individual Plots for (A) ---
for i, K in enumerate(K_values):
    res = results_A[i]

    plt.figure(figsize=(8,5))
    plt.plot(res["t"], res["aM"])
    plt.title(f"(A) Missile Lateral Acceleration vs Time (K={K})")
    plt.xlabel("Time (s)")
    plt.ylabel("aM (m/s²)")
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(8,5))
    plt.plot(res["xM"], res["yM"], label="Missile")
    plt.plot(res["xT"], res["yT"], '--', label="Target")
    plt.title(f"(A) Trajectories (K={K})")
    plt.xlabel("x (m)")
    plt.ylabel("y (m)")
    plt.legend()
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(8,5))
    plt.plot(res["t"], res["RM_theta_diff"])
    plt.title(f"(A) RM(θT − θM) vs Time (K={K})")
    plt.xlabel("Time (s)")
    plt.ylabel("RM·(θT−θM) (m·rad)")
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(8,5))
    plt.plot(res["t"], res["RM"])
    plt.title(f"(A) Separation R vs Time (K={K})")
    plt.xlabel("Time (s)")
    plt.ylabel("R (m)")
    plt.grid(True)
    plt.show()
```

```python
K = 10
xM, yM = xM0, yM0
xT, yT = xT0, yT0
thetaM = thetaM0
thetaT = thetaT0
t = 0

RM_list, aM_list, t_list = [], [], []
xM_list, yM_list, xT_list, yT_list = [], [], [], []

while t < T_max:   # use T_max consistently
    dx = xT - xM
    dy = yT - yM
    RM = np.hypot(dx, dy)
    theta_diff = (thetaT - thetaM)
    theta_diff = (theta_diff + np.pi) % (2 * np.pi) - np.pi

    # Missile lateral acceleration
    aM = K * RM * theta_diff

    # Target constant lateral acceleration
    aT = aT_const

    # Update headings
    thetaM += (aM / VM) * dt
    thetaT += (aT / VT) * dt if VT != 0 else 0

    # Update positions
    xM += VM * np.cos(thetaM) * dt
    yM += VM * np.sin(thetaM) * dt
    xT += VT * np.cos(thetaT) * dt
    yT += VT * np.sin(thetaT) * dt

    # Store
    RM_list.append(RM)
    aM_list.append(aM)
    t_list.append(t)
    xM_list.append(xM)
    yM_list.append(yM)
    xT_list.append(xT)
    yT_list.append(yT)

    if RM < 1:
        break

    t += dt

miss_distance = RM_list[-1]
print(f"(B) K=10, aT=3g: Miss distance = {miss_distance:.2f} m")

# --- Individual Plots for (B) ---
plt.figure(figsize=(8,5))
plt.plot(t_list, aM_list)
plt.title("(B) Missile Lateral Acceleration vs Time")
plt.xlabel("Time (s)")
plt.ylabel("aM (m/s²)")
plt.grid(True)
plt.show()

plt.figure(figsize=(8,5))
plt.plot(xM_list, yM_list, label="Missile")
plt.plot(xT_list, yT_list, '--', label="Target")
plt.title("(B) Trajectories")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(8,5))
plt.plot(t_list, RM_list)
plt.title("(B) Separation R vs Time")
plt.xlabel("Time (s)")
plt.ylabel("R (m)")
plt.grid(True)
plt.show()
```
